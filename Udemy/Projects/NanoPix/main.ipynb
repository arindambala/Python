#### **Conversion : Image to *Black & White* / *Greyscale* - using Luminance Coefficients**

The standard linear sRGB luminance formula defined as :

**$$Y_{linear} = 0.2126 \, R_{linear} + 0.7152 \, G_{linear} + 0.0722 \, B_{linear}$$**

```python
arr_img = img / 255
greyscale = np.array([0.2126, 0.7152, 0.0722])

img_bw = arr_img @ greyscale

plt.imshow(img_bw, cmap='gray')

plt.show()
```

```python
xarr_img = img / 255
greyscale = np.array([0.2126, 0.7152, 0.0722])

ximg_bw = xarr_img @ greyscale

plt.imshow(img_bw)

plt.show()
```

```python
rev_img = img / 255
greyscale = np.array([0.2126, 0.7152, 0.0722])

rev_pix = rev_img @ greyscale

plt.imshow(np.flip(rev_pix), cmap='gray')

plt.show()
```
