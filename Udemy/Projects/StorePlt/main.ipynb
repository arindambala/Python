```python
import pandas as pd
import plotly.express as px

df_apps = pd.read_csv('apps.csv')
```

```python
df_apps.shape
```

```python
df_apps.columns
```

```python
df_apps.head()
```

```python
df_apps.drop(['Last_Updated', 'Android_Ver'], axis=1, inplace=True)
df_apps.head()
```

```python
nan_rows = df_apps[df_apps.Rating.isna()]
print(nan_rows.shape)

nan_rows.head()
```

```python
df_apps_clean = df_apps.dropna()
df_apps_clean.shape
```

```python
duplicated_rows = df_apps_clean[df_apps_clean.duplicated()]
print(duplicated_rows.shape)

duplicated_rows.head()
```

```python
df_apps_clean[df_apps_clean.App == 'Instagram']
```

```python
df_apps_clean = df_apps_clean.drop_duplicates(subset=['App', 'Type', 'Price'])
df_apps_clean[df_apps_clean.App == 'Instagram']
```

```python
df_apps_clean.shape
```

```python
ratings = df_apps_clean.Content_Rating.value_counts()
```

```python
ratings
```

```python
fig = px.pie(labels=ratings.index, values=ratings.values)
fig.show()
```

```python
fig = px.pie(labels=ratings.index, values=ratings.values, title='Content Rating', names=ratings.index)

fig.update_traces(textposition='outside', textinfo='percent+label')

fig.show()
```

```python
fig = px.pie(labels=ratings.index, values=ratings.values, title='Content Rating', names=ratings.index, hole=0.6)

fig.update_traces(textposition='inside', textinfo='percent', textfont_size=15)

fig.show()
```

```python
df_apps_clean.Installs.describe()
```

```python
df_apps_clean.Installs = df_apps_clean.Installs.astype(str).str.replace(',',"")
df_apps_clean.Installs = pd.to_numeric(df_apps_clean.Installs)

df_apps_clean[['App', 'Installs']].groupby('Installs').count()
```

```python
df_apps_clean.Category.nunique()
```

```python
top10_category = df_apps_clean.Category.value_counts()[:10]
```

```python
bar = px.bar(x = top10_category.index, y = top10_category.values)
bar.show()
```

```python
category_installs = df_apps_clean.groupby('Category').agg({'Installs': pd.Series.sum})
category_installs.sort_values('Installs', ascending=True, inplace=True)
```

```python
h_bar = px.bar(x = category_installs.Installs, y = category_installs.index, orientation='h', title='Category Popularity')

h_bar.update_layout(xaxis_title='Total Downloads', yaxis_title='Category')

h_bar.show()
```

```python
scat_num = df_apps_clean.groupby('Category').agg({'App': pd.Series.count})

scat_merged_df = pd.merge(scat_num, category_installs, on='Category', how='inner')
print(f'DataFrame Dimensions : {scat_merged_df.shape}')

scat_merged_df.sort_values('Installs', ascending=False)
```

```python
scatter = px.scatter(scat_merged_df, x = 'App', y = 'Installs', title = 'Category Concentration', size = 'App', hover_name = scat_merged_df.index, color = 'Installs')

scatter.update_layout(xaxis_title = 'Total Apps (Lower = More Concentrated)', yaxis_title = 'Installs', yaxis = dict(type = 'log'))

scatter.show()
```

```python
df_free_vs_paid = df_apps_clean.groupby(['Category', 'Type'], as_index=False).agg({'App': pd.Series.count})
df_free_vs_paid.head()
```

```python
g_bar = px.bar(df_free_vs_paid, x = 'Category', y = 'App', title = 'Free v/s Paid Apps by Category', color = 'Type', barmode = 'group')

g_bar.update_layout(xaxis_title = 'Category', yaxis_title = 'Total Apps', xaxis = {'categoryorder': 'total descending'}, yaxis = dict(type='log'))

g_bar.show()
```

```python
box = px.box(df_apps_clean, x = 'Type', y = 'Installs', color = 'Type', notched = True, points = 'all', title = 'Total downloads paid apps give up are?')

box.update_layout(yaxis = dict(type='log'))

box.show()
```

```python
df_apps_clean['Installs'] = df_apps_clean['Installs'].astype(str).str.replace(',', '').astype(int)
df_apps_clean['Price'] = df_apps_clean['Price'].astype(str).str.replace('$', '').astype(float)

df_apps_clean['Revenue_Estimate'] = df_apps_clean['Installs'] * df_apps_clean['Price']
```

```python
df_apps_paid = df_apps_clean[df_apps_clean.Type == 'Paid']
box = px.box(df_apps_paid,
            y='Revenue_Estimate',
            x='Category',
            title='How much can paid apps earn?',
            )

box.update_layout(yaxis=dict(type='log'),
                xaxis_title='Category',
                yaxis_title='Paid App Ballpark Revenue',
                xaxis_tickangle=30,
                xaxis={'categoryorder':'min ascending'})

box.show()
```

```python
box = px.box(df_apps_paid,
            y='Price',
            x='Category',
            title='Price per Category',
            )

box.update_layout(yaxis=dict(type='log'),
                xaxis_title='Category',
                yaxis_title='Paid App Price',
                xaxis_tickangle=30,
                xaxis={'categoryorder':'max descending'})

box.show()
```
